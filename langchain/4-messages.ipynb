### Messages

Messages are the fundamental unit of context for models in LangChain. they
   - represent input and output of models
   - carry both the content and metadata needed to represent the state of the conversation
Messages are objects that contain:
   - Roles: identifies the message type(e.g System or User)
   - Content: Represents the actual content of the message(like text, image, video..)
   - Metadata: Optional fields such as response information, message IDs, and token usage


##### Langchain Provides a standard message type that works across all model providers, ensuring consistent behavior regardless of the model being called.

```python
import os

from langchain.chat_models import init_chat_model


os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model = init_chat_model("groq:openai/gpt-oss-safeguard-20b")

```

```python
model.invoke("Tell me about the ancient country Ethipia.")
```

### Text Prompts 
Text prompts are strings - ideal for straightforward generation tasks where you don't need to retain conversation history.

```python
model.invoke("What is langgraph?")
```


Use Text prompts when:
  - you have a single, standalone request
  - you don't need conversation history
  - You want minimal code complexity

### Message Prompts
Alternatively, you can pass in a list of messages to the model by providing a list of message objects.

Message types
   - System message - Tells the model how to behave and provide context for interactions
   - Humman Message - Represents user input and interactions with the model
   - AI Message - Responses generated by the model, including text content, tool calls, and metadata
   - Tool Message - Represents the output of tool calls

```python
from langchain.messages import SystemMessage, AIMessage, HumanMessage, ToolMessage

messages = [
    SystemMessage("You are the poetry expert!"),
    HumanMessage("Write a poem on Ethioia")
]

response = model.invoke(messages)
response
```

```python
system_msg = SystemMessage("You are a helpful coding assistant.")

messages = [
    system_msg,
    HumanMessage("How do I create a REST API?")
]
response = model.invoke(messages)
response.content
```
