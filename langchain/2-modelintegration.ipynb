### Google Gemini Model Integration

```python
import os

from langchain.chat_models import init_chat_model


os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
model = init_chat_model("google_genai:gemini-3.6-flash")

response = model.invoke("Why do parrots talk?")
response.content
```

```python
### Google Gen AI

from langchain_google_genai import ChatGoogleGenerativeAI

model_genai = ChatGoogleGenerativeAI(model = "gemini-3.8-flash")
response = model_genai.invoke("Why do parrots talk?")
response.content
```

### GROQ Model Integration

```python
import os

from langchain.chat_models import init_chat_model


os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model = init_chat_model("groq:openai/gpt-oss-safeguard-20b")

response = model.invoke("Why do parrots talk?")
response
```

```python
from langchain_groq import ChatGroq

model_groq = ChatGroq(model ="openai/gpt-oss-safeguard-20b" )
response = model_groq.invoke("why do parrots talk?")
response
```

### Streaming and Batch

### Streaming

```python
response = model_groq.invoke("Write me a 200 words paragraph on Artificial intelligence.")
response.content
```

```python
for chunk in model_groq.stream("Write me a 200 words paragraph on Artificial intelligence."):
    print(chunk.text, end='', flush=True)
```

```python
for chunk in model_genai.stream("Write me a 200 words paragraph on Artificial intelligence."):
    print(chunk.text, end='|', flush=True)
```

```python
for chunk in model_genai.stream("Why do parrots have colorful feathers?"):
    print(chunk.text, end='|', flush=True)
```

### Batch

```python
responses = model_genai.batch([
    "Why do parrots have colorfull feathers? ",
    "How do airplanes fly?",
    "what is quantum computing?"
])

for response in responses:
    print(response)
```
