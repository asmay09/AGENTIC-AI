
### Python OOP

```python
class Person:

    def __init__(self, name, age, job):
        self.name = name
        self.age = age
        self.job = job

    def __str__(self):
        return f'Person <Name: {self.name}>'

    def print_name(self):
        print(f'Name is {self.name}')

    def __repr__(self):
        return f"Name: {self.name}, Age: {self.age}, Job: {self.job}"

    def getOlder(self):
        self.age +=1
```

```python

person1 = Person("Asmare", 26, "Programmer")

print(person1.name)
print(person1.age)
print(person1.job)

person1.getOlder()
person1.age
print(person1)
```

### Inheritance

```python
class Programmer(Person):

    def __init__(self, name, age, job, main_language):
        super().__init__(name, age, job)
        self.main_language = main_language

    def __repr__(self):
        return f"Name: {self.name}, Age: {self.age}, Job: {self.job}, Language: {self.main_language}"
```

```python
prog1 = Programmer("Mike", 35, "Programmmer", "C++")

prog1.print_name()
print(prog1.main_language)
print(prog1)
```
